```python
import pandas as pd

df = pd.read_csv("data/combined_multilingual_mmlu.csv")

df.head()
```

```python
df.columns
```

```python
def format_for_training(row):
    return {
        "text": f"""### Instruction:
Answer the following multiple choice question and explain your reasoning clearly.

### Input:
Question: {row['Question']}

A: {row['A']}
B: {row['B']}
C: {row['C']}
D: {row['D']}

### Response:
Correct Answer: {row['Answer']}
Explanation:"""
    }

df_formatted = df.apply(format_for_training, axis=1, result_type="expand")
df_formatted.head()
```

```python
import torch
from transformers import AutoModelForCausalLM, AutoTokenizer

model_path = "model/"

tokenizer = AutoTokenizer.from_pretrained(model_path)

model = AutoModelForCausalLM.from_pretrained(
    model_path,
    torch_dtype=torch.bfloat16,  # better stability than fp16
    device_map="cuda"            # forces full model to GPU
)

model.config.use_cache = False
model.eval()
```

```python
from datasets import Dataset

hf_dataset = Dataset.from_pandas(df_formatted)
hf_dataset = hf_dataset.train_test_split(test_size=0.1)
```

```python
def tokenize(example):
    return tokenizer(
        example["text"],
        truncation=True,
        max_length=1024,
        padding="max_length"
    )

tokenized_dataset = hf_dataset.map(tokenize, batched=True)
```

```python
from peft import LoraConfig, get_peft_model

lora_config = LoraConfig(
    r=16,
    lora_alpha=32,
    target_modules=["q_proj","v_proj"],
    lora_dropout=0.05,
    bias="none",
    task_type="CAUSAL_LM",
)

model = get_peft_model(model, lora_config)
model.print_trainable_parameters()
```

```python
from transformers import TrainingArguments
from trl import SFTTrainer

training_args = TrainingArguments(
    output_dir="./deepseek-xai",
    per_device_train_batch_size=4,
    gradient_accumulation_steps=4,
    num_train_epochs=2,
    learning_rate=2e-4,
    logging_steps=50,
    save_steps=500,
    fp16=True,
    report_to="none"
)
```

```python
from trl import SFTTrainer, SFTConfig

sft_config = SFTConfig(
    output_dir="./deepseek-xai",
    per_device_train_batch_size=4,
    gradient_accumulation_steps=4,
    num_train_epochs=2,
    learning_rate=2e-4,
    logging_steps=10,
    eval_strategy="steps",
    eval_steps=100,
    save_steps=500,
    bf16=True,
    report_to="none",
)

trainer = SFTTrainer(
    model=model,
    train_dataset=hf_dataset["train"],
    eval_dataset=hf_dataset["test"],
    processing_class=tokenizer,
    args=sft_config,
)

trainer.train()


```
